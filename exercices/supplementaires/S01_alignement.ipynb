# S01 — Alignement des Series

## Énoncé

Deux entrepôts publient leur stock sous forme de Series indexées par référence produit. Le service achats additionne les deux Series pour obtenir le stock total et trouve un résultat incohérent. Il faut comprendre pourquoi et produire un total correct.

## Règles et étapes

1. Pour chacune des quatre expressions de la cellule « Prédictions », écrire **avant d'exécuter** le résultat attendu (valeurs, index, type).
2. Exécuter et comparer avec vos prédictions.
3. Calculer le stock total correct, en considérant qu'une référence absente d'un entrepôt y a un stock nul. Le résultat doit être de type entier.
4. Lister les références présentes dans un seul des deux entrepôts.

## Objectifs pédagogiques

- Prévoir l'effet de l'alignement par l'index.
- Expliquer l'apparition des `NaN` et le passage en `float64`.
- Utiliser `add(..., fill_value=...)` et `astype`.

## Mockup (forme du résultat attendu)

```
P01    32
P02    40
P03     7
...
dtype: int64
```

## Contraintes

- Pas de boucle `for`.
- Ne pas modifier les Series de départ.

````python
import pandas as pd

nord = pd.Series({"P01": 12, "P02": 40, "P03": 7, "P05": 3})
sud = pd.Series({"P01": 20, "P04": 15, "P05": 0, "P06": 8})
````

### Prédictions

```python
nord + sud                       # (a)
(nord + sud).sum()               # (b)
nord.values + sud.values         # (c)
nord.add(sud, fill_value=0)      # (d)
```

````python
print(nord + sud, end="\n\n")
print((nord + sud).sum(), end="\n\n")
print(nord.values + sud.values, end="\n\n")
print(nord.add(sud, fill_value=0))
````

````python
total = nord.add(sud, fill_value=0).astype(int)
total
````

````python
seulement_un = nord.index.symmetric_difference(sud.index)
list(seulement_un)
````
