# S05 — Les modifications qui ne modifient rien

## Énoncé

Le script de nettoyage ci-dessous a été écrit pour une ancienne version de Pandas. En Pandas 3, il s'exécute presque sans erreur, mais une partie des modifications n'a aucun effet. Il faut identifier lesquelles et réécrire le script.

## Règles et étapes

1. Exécuter la cellule « script d'origine », puis la cellule de contrôle.
2. Pour chacune des quatre opérations, dire si elle a modifié `df`.
3. Réécrire le script pour que les quatre opérations soient effectives, sans utiliser `inplace=True`.
4. Relancer la cellule de contrôle sur votre version.

## Objectifs pédagogiques

- Reconnaître une affectation chaînée.
- Appliquer les deux formes correctes : `df["col"] = ...` et `df.loc[lignes, col] = ...`.
- Comprendre le principe du Copy-on-Write.

## Mockup

```
statut manquants : 0
canal en minuscules : True
quantité des annulées à 0 : True
prix rempli : True
```

## Contraintes

- Ne pas utiliser `inplace=True`.
- Une seule instruction par opération.

````python
import pandas as pd

def charger():
    brut = pd.read_csv("../../donnees/demo/commandes_brutes.csv", sep=";").drop_duplicates("commande_id")
    brut["quantite"] = pd.to_numeric(brut["quantite"], errors="coerce")
    return brut

def controle(d):
    print("statut manquants :", d["statut"].isna().sum())
    print("canal en minuscules :", (d["canal"] == d["canal"].str.strip().str.lower()).all())
    print("quantité des annulées à 0 :", (d.loc[d["statut"] == "annulée", "quantite"] == 0).all())
    print("prix rempli :", d["prix_unitaire"].notna().all())
````

````python
# Script d'origine
df = charger()
df["statut"].fillna("inconnu", inplace=True)                         # 1
df["canal"] = df["canal"].str.strip().str.lower()                    # 2
df[df["statut"] == "annulée"]["quantite"] = 0                        # 3
df.prix_unitaire.fillna("0,00 €", inplace=True)                      # 4
print("script terminé")
````

````python
controle(df)
````
