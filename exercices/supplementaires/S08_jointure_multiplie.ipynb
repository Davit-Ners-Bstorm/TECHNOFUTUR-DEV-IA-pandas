# S08 — La jointure qui multiplie les lignes

## Énoncé

Pour calculer le chiffre d'affaires, on joint les commandes au fichier des tarifs. Le total obtenu dépasse nettement celui du mois précédent, sans raison commerciale. Le fichier des tarifs contient l'historique des prix : un produit peut y apparaître plusieurs fois.

## Règles et étapes

1. Calculer le chiffre d'affaires de référence avec la table `produits` (un prix par produit).
2. Calculer le chiffre d'affaires avec la table `tarifs`, et comparer le nombre de lignes avant et après la jointure.
3. Ajouter le paramètre `validate` adapté à la jointure et observer le message.
4. Corriger : pour chaque commande, retenir le tarif **en vigueur à la date de la commande** (date de début la plus récente antérieure ou égale à la date de commande).
