# S10 — Remplacer `apply` par du code vectorisé

## Énoncé

Une fonction attribue à chaque participation olympique une catégorie d'âge, et un indicateur « médaillé ». Appliquée avec `apply` sur 271 116 lignes, elle est lente. Il faut obtenir exactement le même résultat avec des opérations vectorisées, et mesurer le gain.

## Règles et étapes

1. Exécuter la version `apply` et mesurer son temps avec `%timeit`.
2. Écrire une version vectorisée de `categorie_age` avec `np.select` (ou `pd.cut`).
3. Écrire une version vectorisée de `points` (or = 3, argent = 2, bronze = 1, sinon 0) avec `map`.
4. Vérifier que les résultats sont **identiques** aux versions `apply`, y compris pour les âges manquants.
5. Mesurer et comparer les temps.

## Objectifs pédagogiques

- Traduire une logique conditionnelle en `np.select`.
- Vérifier l'équivalence de deux implémentations.
- Mesurer avec `%timeit`.

## Contraintes

- Pas de `apply`, `map` avec fonction, ni de boucle dans la version vectorisée (`map` avec un dictionnaire est autorisé).

````python
import numpy as np
import pandas as pd
jo = pd.read_csv("../../donnees/bstorm/Exercices_Pandas_athlete_events.csv")

def categorie_age(age):
    if pd.isna(age):
        return "inconnu"
    if age < 18:
        return "junior"
    if age < 35:
        return "senior"
    return "master"

def points(medaille):
    if medaille == "Gold":
        return 3
    if medaille == "Silver":
        return 2
    if medaille == "Bronze":
        return 1
    return 0
````

````python
%timeit jo["Age"].apply(categorie_age)
%timeit jo["Medal"].apply(points)
````
