```python
in_cold_war = (data['Year'] >= 1952) & (data['Year'] <= 1988)
```

```python
GB = data.loc[in_cold_war,['NOC','Name', 'Medal']].groupby('NOC').count()
```

```python
GB.sort_values(by= 'Name',ascending=False)
```

```python
GB['eff'] = GB['Medal']/GB['Name']
display(GB.sort_values(by= 'Medal',ascending=False).head(10))
display(GB.sort_values(by= ['eff','Medal'],ascending=False).head(10))
```

> **[Ajout] Remarque sur la question 32.** `count()` sur `Name` compte des **participations** (un athlète présent à trois épreuves compte trois fois) et `count()` sur `Medal` compte des **médaillés**. Variante en athlètes distincts et en médailles (voir le chapitre 7 du cours) :

```python
periode = data[in_cold_war]
eff = pd.DataFrame({
    'athletes': periode.groupby('NOC')['ID'].nunique(),
    'medailles': periode.dropna(subset=['Medal']).drop_duplicates(subset=['Event', 'Games', 'NOC', 'Medal']).groupby('NOC').size(),
}).fillna(0)
eff['medailles_par_athlete'] = eff['medailles'] / eff['athletes']
eff[eff['athletes'] >= 100].sort_values('medailles_par_athlete', ascending=False).head(10)
```

# .info cusom:

```python
def INFO(data):
    res = pd.concat([
        data.dtypes
        ,data.isnull().sum()
        ,data.nunique()
    ],axis=1)
    res.columns = ['dtype','isnull_sum','nunique']
    
    return res
```
