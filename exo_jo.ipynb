```python
#31)	Quels sont les pays gagnant les plus médaillés par sport distinct
medals = data.groupby(['Sport', 'NOC'])['Medal'].count().reset_index(name='nb_medailles')
 
# Pour chaque sport, garder le pays qui a le maximum
top = medals.loc[medals.groupby('Sport')['nb_medailles'].idxmax()]
top = top.sort_values('nb_medailles', ascending=False)
top
```

```python
periode = data[data["Year"].between(1952, 1988)]
medailles_periode = medailles[medailles["Year"].between(1952, 1988)]

efficacite = pd.DataFrame({
    "athletes": periode.groupby('NOC')["ID"].nunique(),
    "medailles": medailles_periode.groupby('NOC').size()
}).fillna(0)

efficacite["medailles_par_athlete"] = (efficacite["medailles"] / efficacite["athletes"]).round(3)

efficacite.sort_values("medailles_par_athlete", ascending=False).head()
```

```python
medailles_par_pays = medailles["NOC"].value_counts()
periode = data.loc[data["Year"].between(1952, 1988)]
athlete_par_pays = periode.groupby("NOC")["ID"].nunique()
par_pays = medailles.groupby("NOC").size()
efficacite = (medailles_par_pays / athlete_par_pays).fillna(0)
print(efficacite.sort_values(ascending=False).head().round(3))
```
